# Numeric features

Aggregate training-split feature summaries only. Labels and provenance never enter the feature interface. Requires the optional ML environment. No test data is inspected here. Context and format heuristics may encode synthetic-template shortcuts; importance is not causal evidence.

```python
import os
import sys
from pathlib import Path

root = Path.cwd()
while not (root / "ml" / "data").exists() and root != root.parent:
    root = root.parent

if str(root) not in sys.path:
    sys.path.insert(0, str(root))
```

```python
from statistics import mean

from secretsense.features import FEATURE_NAMES

from ml.scripts.train_baseline import features, read_dataset

rows, _ = read_dataset(Path(os.environ.get("SECRETSENSE_DATA_DIR", root / "ml" / "data")))
train = [row for row in rows if row["split"] == "train"]
for label in ("0", "1"):
    vectors = features([row for row in train if row["label"] == label])
    print("Training label", label, "rows", len(vectors))
    for name, column in zip(FEATURE_NAMES, zip(*vectors, strict=True), strict=True):
        print(
            name,
            "min/mean/max",
            round(min(column), 4),
            round(mean(column), 4),
            round(max(column), 4),
        )
```
